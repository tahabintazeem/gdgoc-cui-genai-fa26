# 🗺️ Map of Generative AI

### Theory 1: What "Generative" Means
*   **Discriminative (The Art Critic):** Categorizes existing data (e.g., "This is a dog").
*   **Generative (The Artist):** Creates net-new high-dimensional data from a prompt (e.g., generating an image or text).

### Theory 2: Neural Networks & Timeline
*   **Neural Networks:** Like a "Real Estate Committee". Different layers check different features, weigh them, and output a result.
*   **RNNs:** Read sequentially (word-by-word) but have short-term memory loss.
*   **Transformers:** Solved this with "Self-Attention" (looking at the whole sentence at once).
*   **Diffusion:** Generates images by removing static/noise from an old TV until a clear picture remains.

```python
# Cell 2: The Art Critic (Discriminative AI)
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
result = classifier("I am incredibly excited for this GDGoC GenAI sprint!")

print(f"Discriminative Output: {result}")
# 💡 TRY THIS: Change the text to something negative and see how the label changes.
```

```python
# Cell 3: The Artist (Generative AI)
# Note: This might take a few seconds to download the model the first time.
generator = pipeline("text-generation", model="gpt2")
prompt = "The most powerful feature of artificial intelligence is"

gen_result = generator(prompt, max_length=30, num_return_sequences=1, temperature=0.7)

print(f"Generative Output: \n{gen_result[0]['generated_text']}")
# 💡 TRY THIS: Change the prompt, increase max_length to 50, and set temperature to 1.5!
```

## 🧠 Takeaways & Insights
1.  **Probability, not magic:** Generative AI simply predicts the next most mathematically logical token/pixel based on its training.
2.  **Accessible:** You don't need massive hardware. You can run open-source models (like GPT-2) in Colab for free.
3.  **Architecture > Frameworks:** Frameworks (like LangChain) change, but understanding architectures (like Transformers) will keep you relevant.
